```python

import numpy as np

def calculate_paid_scaled(
    triangle,
    ldf,
    tail,
    scale_m=1.0,
    scale_a=0.0,
    inflacja=None,
    discount_factors=None,
    start_year=2003
):
    C = np.asarray(triangle, dtype=np.float64)
    a = C.shape[0]
    end_year = start_year + a - 1

    scale_m = np.asarray(scale_m, dtype=np.float64)
    scale_a = np.asarray(scale_a, dtype=np.float64)

    # Parametr wspólny albo osobny dla każdego roku szkodowego
    if scale_m.ndim == 0:
        scale_m = np.full(a, float(scale_m))
    if scale_a.ndim == 0:
        scale_a = np.full(a, float(scale_a))

    if scale_m.shape != (a,) or scale_a.shape != (a,):
        raise ValueError("scale_m i scale_a muszą być skalarami lub wektorami długości a.")

    # 1. Trójkąt wypłat bez przyszłej inflacji
    paid_raw, _ = calculate_paid_fit(
        triangle=C,
        ldf=ldf,
        tail=tail,
        inflacja=None,
        discount_factors=None,
        start_year=start_year
    )

    # 2. Maska przyszłych wypłat
    years = (
        start_year
        + np.arange(a)[:, None]
        + np.arange(a + 1)[None, :]
    )
    future_mask = years > end_year

    # 3. Skalowanie wyłącznie przyszłych wypłat
    paid_scaled = paid_raw.copy()

    scaled_values = (
        scale_m[:, None] * paid_raw
        + scale_a[:, None]
    )

    paid_scaled[future_mask] = scaled_values[future_mask]

    # 4. Inflacja przyszłych wypłat
    if inflacja is not None:
        inf = np.asarray(inflacja, dtype=np.float64)
        inflation_dict = {
            int(year): float(rate)
            for year, rate in inf
        }

        for i, j in np.argwhere(future_mask):
            year = int(years[i, j])

            factor = 1.0
            for y in range(end_year + 1, year):
                factor *= 1.0 + inflation_dict[y]

            factor *= np.sqrt(1.0 + inflation_dict[year])

            paid_scaled[i, j] *= factor

    # 5. BEL
    if discount_factors is None:
        df = np.ones_like(paid_scaled)
    else:
        factors = np.asarray(discount_factors, dtype=np.float64)
        offsets = years - end_year - 1

        required = np.max(offsets[future_mask]) + 1
        if len(factors) < required:
            raise ValueError("Za krótki wektor discount_factors.")

        df = np.ones_like(paid_scaled)
        df[future_mask] = factors[offsets[future_mask]]

    BEL = float(np.sum(
        np.where(future_mask, paid_scaled * df, 0.0)
    ))

    return paid_scaled, BEL
```
